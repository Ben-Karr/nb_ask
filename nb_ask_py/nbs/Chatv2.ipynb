```python
from fastllm.chat import AsyncChat
import httpx
```

```python
EXT_URL = "http://localhost"
EXT_PORT = "3144"

def get_ext_config():
    """Fetch token/model from the VS Code extension; fall back to .env."""
    ret = httpx.get(f"{EXT_URL}:{EXT_PORT}/config", timeout=2)
    ret.raise_for_status()
    cfg = ret.json()
    return cfg.get("token"), cfg.get("endpoint"), cfg.get("model")


key, endpoint, full_model = get_ext_config()
model = full_model.removeprefix("openrouter/")
```

```python
from cellm.core import sp
from cellm.tools import *
from fastcore.tools import view_file
```

```python
chat = AsyncChat(
    model=model,
    sp=sp,
    tools=[web_search, web_fetch, tree, grep, view_file],
    vendor_name='openrouter',
    api_key=key,
    base_url=endpoint,   # keeps your ext-config endpoint as override; drop if it's just openrouter's URL
)
```

```python
r = httpx.get("https://openrouter.ai/api/v1/models", timeout=10)
models_res = r.json()
```

```python
from fastllm.chat import register_model_info
```

```python
model_info = next(x for x in models_res["data"] if x['id'] == model)
register_model_info(model, vendor_name='openrouter',
    input_cost_per_token=float(model_info['pricing']['prompt']),
    output_cost_per_token=float(model_info['pricing']['completion']),
    cache_read_input_token_cost=float(model_info['pricing'].get('prompt_cache_read', 0)) or 0,
    max_output_tokens=32_000)
```

```python
pr = "Whats the current weather in France?"
```

```python
l = []
async for chunk in await chat(pr, stream=True):
    l.append(chunk)
```

```python
from IPython.display import Markdown, display
import json
from fastllm.types import Part, PartType, Completion


handle = display(Markdown("⏳ …"), display_id=True)   # returns a DisplayHandle
md = []                                                # accumulated markdown

async for chunk in await chat(pr, stream=True, max_steps=6):
    if isinstance(chunk, dict):                        # streamed text/thinking
        if chunk.get('thinking'): md.append(f"> 🧠 {chunk['thinking']}")
        if chunk.get('text'):     md.append(chunk['text'])
    elif isinstance(chunk, Part):                      # tool event
        d = chunk.data
        if chunk.type == PartType.tool_use:
            md.append(f"🔧 **{d['name']}**`({json.dumps(d['arguments'])[:80]})`")
        else:
            md.append(f"↳ `{(chunk.text or '')[:100]}`")
    elif isinstance(chunk, Completion):
        final = chunk

    handle.update(Markdown("\n\n".join(md)))           # re-render the cell output
```

```python
from cellm.core import get_cells

async def ask(pr: str):
    _, cells_raw = get_cells()
    chat.hist = [f"active notebook content: {json.dumps(cells_raw)}"]
    fmt = await adisplay_stream(await chat(pr, stream=True, max_steps=6), showthink=True)
    model_name = chat.model.split('/')[-1].split('/')[-1]  # strip vendor too
    header = f"##### 🤖 LLM Response ({model_name} | {chat.use})\n\n"
    return display(Markdown(header + fmt.outp))
```
